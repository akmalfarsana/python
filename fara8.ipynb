```python
square=lambda a:a*a
triangle=lambda l,h:0.5*l*h 
rectangle=lambda b,h:b*h
while True:
    print("\n1.Square\n2.Triangle\n3.Rectangle\n4.Exit")
    ch=int(input("Enter your choice:"))
    if(ch==1):
        a=int(input("Enter value:"))
        print("Area",square(a))
    elif(ch==2):
        l=int(input("Enter the length:"))
        h=int(input("Enter the height:"))
        print("Area of triangle",triangle(l,h))
    elif(ch==3):
        b=int(input("Enter the base:"))
        h=int(input("Enter the height:"))
        print("Area of rectangle",rectangle(b,h))
    elif(ch==4):
        print("Exit")
        break
    else:
        print("Invalid choice.")
```

```python
lst1=list(map(int,input("Enter first list:").split()))
lst2=list(map(int,input("Enter second list:").split()))
if len(lst1)==len(lst2):
    print("Both lists have the same length")
else:
    print("Both lists have different length")
    
if sum(lst1)==(sum(lst2)):
    print("Both lists have the same sum")
else:
    print("Both lists have the different sum")
    
if set(lst1)&set(lst2):
    print("common values:",set(lst1)&set(lst2))
else:
    print("No common values")
```
